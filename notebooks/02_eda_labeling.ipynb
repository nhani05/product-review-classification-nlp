### Ph?n t?ch ph?n b? ?? d?i v? nh?n
So s?nh ?? d?i c?u v? t? l? spam/ham.

```python
with open('../data/labeled/reviews_labeled.csv', 'r', encoding='utf-8-sig') as f:
    reader = csv.DictReader(f)
    data = list(reader)

labels = [int(r['label']) for r in data]
ratings = [r['rating'] for r in data]
lengths = [len(r['review']) for r in data]

print(f'T?ng s? m?u: {len(data)}')
print(f'T? l? Spam: {labels.count(1) / len(labels) * 100:.2f}%')
print(f'Ph?n b? rating: {Counter(ratings)}')
print(f'?? d?i review trung b?nh: {sum(lengths)/len(lengths):.1f} k? t?')
```
